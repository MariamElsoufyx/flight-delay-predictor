## 4 — Fix Identified Issues in Airports Dataset

Strip leading/trailing whitespace and normalise casing for `AirportName` and `City_Name`.

```python
airports = pd.read_csv('../data/raw/global_airports.csv')

str_cols = airports.select_dtypes(include='object').columns

print('Before cleaning:')
print(f'  Rows: {len(airports)}')
print(f'  Sample AirportName values: {airports["AirportName"].head(5).tolist()}')

# Strip whitespace on all string columns
for col in str_cols:
    airports[col] = airports[col].str.strip()

# Title-case AirportName and City_Name for consistency
airports['AirportName'] = airports['AirportName'].str.title()
airports['City_Name']   = airports['City_Name'].str.title()

# Upper-case IATA and ICAO codes
airports['IATA'] = airports['IATA'].str.upper()
airports['ICAO'] = airports['ICAO'].str.upper()

print('\nAfter cleaning:')
print(f'  Sample AirportName values: {airports["AirportName"].head(5).tolist()}')
print(f'  Remaining whitespace issues: {airports[str_cols].apply(lambda c: c.str.startswith(" ") | c.str.endswith(" ")).any().any()}')

# Save cleaned airports
os.makedirs('../data/processed', exist_ok=True)
airports.to_csv('../data/processed/airports_clean.csv', index=False)
print('\nSaved cleaned airports to data/processed/airports_clean.csv')
```
